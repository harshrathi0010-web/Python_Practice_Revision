```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6]
y=[134,435,653,678,888,1020]
```

```python
plt.figure(figsize=(10,4))
plt.plot(x,y)
plt.title("Sales by month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
x=["A","B","C","D","E"]
y=[100,324,4543,546,324]
```

```python
plt.figure(figsize=(10,4))
plt.bar(x,y)
plt.show()
```

```python
df=pd.DataFrame({
    "Month:":["jan","feb","mar","apr","may"],
    "Sales:":[3423,45345,6454,-565,65675]
})
```

```python
df.head(4)
```

```python
df.info()
```

```python
plt.figure(figsize=(10,4))
plt.pie(df['Sales:'],df['Month:'])
plt.show()
```

```python
categories=["A","B","C","D","E"]
sales=[10,20,30,40,55]
```

```python
plt.figure(figsize=(10,4))
plt.pie(sales,labels=categories,autopct="%1.1f%%",startangle=90)
plt.show()
```
