```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel(r"C:\Users\DELL\Downloads\Financial_Sample.xlsx")
```

```python
df.head()
```

```python
df=df.bfill()
```

```python
df.isnull().sum()
```

```python
# Profit by segment
pbs=df.groupby('Segment')['Profit'].sum().reset_index()
```

```python
pbs
```

```python
plt.figure(figsize=(8,4))
plt.bar(pbs["Segment"],pbs["Profit"])
plt.show()
```

```python
# profit by month
pbm=df.groupby("Month Number").agg({"Profit":"sum"}).reset_index()
```

```python
plt.figure(figsize=(10,4))
plt.bar(pbm["Month Number"],pbm["Profit"])
plt.show()
```

```python
# Profit by Product
pbp=df.groupby("Product").agg({"Profit":"sum"}).reset_index()
```

```python
pbp
```

```python
plt.figure(figsize=(10,4))
plt.pie(pbp["Profit"],labels=pbp["Product"],autopct="%1.1f%%")
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.pie(pbp["Profit"],labels=pbp["Product"],autopct="0.2f%%")
plt.show()
```

```python
pks=df.groupby("Product").agg({"Profit":"sum"}).reset_index()
```

```python
pks
```

```python
plt.figure(figsize=(10,4))
plt.plot(pks["Product"],pks["Profit"])
plt.show()
```
