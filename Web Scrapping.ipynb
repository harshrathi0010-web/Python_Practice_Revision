```python
import requests
from bs4 import BeautifulSoup
```

```python
url="https://github.com/topics"
```

```python
content=requests.get(url)
```

```python
for row in soup.find_all("a",class_="no-underline flex-1 d-flex flex-column"):
    print(row.find('p').text.strip())
```

```python
topics=[]
desc=[]
links=[]
```

```python
for row in soup.find_all("a",class_="no-underline flex-1 d-flex flex-column"):
    topics.append(row.find('p').text.strip())
    
    desc.append(row.find_all('p')[1].text.strip())
    links.append('https://github.com'+row['href'])
```

```python
len(topics)
```

```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame({
    "topics":topics,
    "desc":desc,
    "links":links
})
```
