```python
import pandas as pd
import numpy as np
```

```python
pd.__version__
```

```python
data=({
    "id":[id for id in range(101,111)],
    "name":["harsh","manish","annu","suresh","ankit","sonu","monu","raju","aniket","sunil"],
    "salary":[34354,5354,43534,534534,54353,65677,87876,89854,123434,78976]
})
          
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
round(df["salary"].mean(),0)
```
