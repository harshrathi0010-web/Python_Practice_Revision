# "While Lists offer flexibility and dynamic sizing, we still use Arrays primarily for performance, memory efficiency, and predictability.
Arrays store elements in contiguous memory without any object or resizing overhead, which gives them superior CPU cache locality and zero memory waste. Furthermore, dynamic lists like Java's ArrayList or Python's list are actually implemented using arrays under the hood. So, when the collection size is fixed and high throughput or low memory usage is critical, an array is the better choice."

```python
#2D
arr=np.array([[23,10,500]])
print("value",arr)
print("shape",arr.shape)
print("dimension",arr.ndim)
```

```python
#3D
arr=np.array([[[23,10,500],[1,2,3],[4,5,6],[10,12,12],[50,10,34]]])
print("value",arr)
print("shape",arr.shape)
print("dimension",arr.ndim)
```

```python
print(arr.shape)       # (1, 5, 3)
print(arr[0])          # the whole block: all 5 rows
print(arr[0, 2])       # [4 5 6]  -> third row
print(arr[0, 2, 1])    # 5        -> block 0, row 2, column 1
```

```python
#3D
arr=np.array([[[23,10,500]]])
print("value",arr)
print("shape",arr.shape)
print("dimension",arr.ndim)
```
